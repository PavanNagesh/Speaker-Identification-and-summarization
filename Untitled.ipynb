```python
# Cell 4: Define the PyTorch Dataset Class

# --- Configuration ---
# 3 seconds * 16000 Hz sample rate = 48000 samples
FIXED_LEN_SAMPLES = 48000 
# We need the spectrogram to be 94 frames wide for the CNN
N_FRAMES = 94 

class AudioDataset(Dataset):
    """
    A PyTorch Dataset class to load and process audio files into Mel Spectrograms.
    """
    def __init__(self, df):
        self.df = df
        
    def __len__(self):
        return len(self.df)
    
    def __getitem__(self, idx):
        # 1. Get the file path and label from the DataFrame
        file_path = self.df.loc[idx, 'file_path']
        label = self.df.loc[idx, 'label']
        
        # 2. Load the audio file (16kHz sample rate)
        # We load a bit extra (3.1s) to ensure we get 48000 samples even with rounding
        y, sr = librosa.load(file_path, sr=16000, duration=3.1) 
        
        # 3. Pad or truncate the audio to FIXED_LEN_SAMPLES (48000)
        if len(y) < FIXED_LEN_SAMPLES:
            y = np.pad(y, (0, FIXED_LEN_SAMPLES - len(y)), 'constant')
        else:
            y = y[:FIXED_LEN_SAMPLES]
        
        # 4. Create the Mel Spectrogram "image" (80 mels)
        S_mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=80)
        S_mel_db = librosa.power_to_db(S_mel, ref=np.max)
        
        # 5. We must ensure the shape is exactly (80, 94) for the CNN
        if S_mel_db.shape[1] > N_FRAMES:
            S_mel_db = S_mel_db[:, :N_FRAMES]
        elif S_mel_db.shape[1] < N_FRAMES:
            S_mel_db = np.pad(S_mel_db, ((0, 0), (0, N_FRAMES - S_mel_db.shape[1])), 'constant')

        # 6. Convert to PyTorch Tensors (add channel dimension)
        spectrogram_tensor = torch.tensor(S_mel_db, dtype=torch.float32).unsqueeze(0)
        label_tensor = torch.tensor(label, dtype=torch.long)
        
        return spectrogram_tensor, label_tensor

print("AudioDataset class defined successfully.")
```

```python
# Cell 6: Define the Hybrid CNN-GRU Encoder Model

class SpeakerEncoder(nn.Module):
    def __init__(self, num_speakers):
        """
        Initializes the CNN-GRU hybrid model architecture.
        num_speakers (int): The number of unique speakers (e.g., 251).
        """
        super(SpeakerEncoder, self).__init__()
        
        # --- CNN Layers (Feature Extractor) ---
        # Input: [Batch_Size, 1, 80, 94]
        
        # Conv block 1 -> output shape: [Batch_Size, 32, 40, 47]
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, padding=1)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)

        # Conv block 2 -> output shape: [Batch_Size, 64, 20, 23]
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)
        
        # Conv block 3 -> output shape: [Batch_Size, 128, 10, 11] (10 freq bins, 11 time steps)
        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, padding=1)
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)
        
        # --- Recurrent Layer (Temporal Processor) ---
        # Input features are 128 channels * 10 frequency bins = 1280
        self.gru_input_size = 128 * 10 
        self.gru_hidden_size = 512
        self.gru = nn.GRU(
            input_size=self.gru_input_size, 
            hidden_size=self.gru_hidden_size, 
            num_layers=1, 
            batch_first=True # Expects input as [Batch, Seq_Len, Features]
        )

        # --- Fully Connected Layers (Classifier) ---
        self.embedding_dim = 768 # The voice fingerprint dimension
        
        self.fc1_embedding = nn.Linear(
            in_features=self.gru_hidden_size, 
            out_features=self.embedding_dim 
        )
        
        self.fc2_classifier = nn.Linear(
            in_features=self.embedding_dim, 
            out_features=num_speakers
        )

    def forward(self, x):
        
        # 1. Pass through CNN blocks (Spatial Feature Extraction)
        # x shape starts as [B, 1, 80, 94]
        x = F.relu(self.conv1(x))
        x = self.pool1(x)
        
        x = F.relu(self.conv2(x))
        x = self.pool2(x)
        
        x = F.relu(self.conv3(x))
        x = self.pool3(x) # x shape is [B, 128, 10, 11]
        
        # 2. Reshape for GRU (Prepare for Temporal Sequence Processing)
        # We want [B, 11 (Time), 1280 (Features)]
        x = x.permute(0, 3, 2, 1) # x shape: [B, 11 (time), 10 (freq), 128 (channels)]
        x = x.reshape(x.size(0), x.size(1), -1) # x shape: [B, 11, 1280]
        
        # 3. Pass through GRU (Temporal Aggregation)
        # We use the final hidden state (hidden) which summarizes the entire sequence
        gru_output, hidden = self.gru(x)
        x = hidden.squeeze(0) # x shape: [B, 512]

        # 4. Get the 768-dim embedding (Voice Fingerprint)
        x_embedding = F.relu(self.fc1_embedding(x))
        
        # 5. Get the final classification output (Logits)
        output_logits = self.fc2_classifier(x_embedding)
        
        return output_logits

print("SpeakerEncoder (CNN-GRU Hybrid) class defined successfully.")
```

```python
# Cell 7: Instantiate Model, Loss, and Optimizer

# --- 1. Define Device ---
# 'device' variable was set in Cell 1
print(f"Using device: {device} ({torch.cuda.get_device_name(0)})")

# --- 2. Instantiate Model and Move to GPU ---
# num_speakers variable (251) comes from Cell 3
model = SpeakerEncoder(num_speakers=num_speakers)
model.to(device)

# --- 3. Define Loss Function ---
criterion = nn.CrossEntropyLoss()

# --- 4. Define Optimizer ---
# Using the original, stable learning rate
optimizer = optim.Adam(model.parameters(), lr=0.0001) 

print("Model, Criterion, and Optimizer are all set up.")

# --- 5. Final Test Pass on GPU ---
print("Testing model with one batch on GPU...")
try:
    # Get the test batch from Cell 5
    test_batch_gpu = one_batch_spectrograms.to(device)
    
    # Perform a forward pass
    with torch.no_grad(): # We don't need to calculate gradients for this test
        output_logits = model(test_batch_gpu)
    
    # Expected output shape: [Batch Size (64), Num Speakers (251)]
    print(f"✅ Success! Model ran on GPU.")
    print(f"   Output shape: {output_logits.shape} (Batch Size, Num Speakers)")

except Exception as e:
    print(f"❌ An error occurred during the model test: {e}")
```

```python
# Cell 8: Define Training and Validation Functions

# --- Define the function for one training epoch ---
def train_one_epoch(model, data_loader, criterion, optimizer, device):
    model.train() # Set the model to "training mode"
    
    total_loss = 0.0
    correct_predictions = 0
    total_samples = 0
    
    # Use tqdm for a progress bar over our data loader
    for spectrograms, labels in tqdm(data_loader, desc="Training", leave=False):
        # Move data to the GPU
        spectrograms = spectrograms.to(device)
        labels = labels.to(device)
        
        # 1. Clear old gradients
        optimizer.zero_grad()
        
        # 2. Forward pass
        outputs = model(spectrograms)
        
        # 3. Calculate the loss
        loss = criterion(outputs, labels)
        
        # 4. Backward pass: Calculate gradients
        loss.backward()
        
        # 5. Optimizer step: Update model weights
        optimizer.step()
        
        # --- Calculate running statistics ---
        total_loss += loss.item() * spectrograms.size(0)
        _, predictions = torch.max(outputs, 1)
        correct_predictions += (predictions == labels).sum().item()
        total_samples += labels.size(0)
        
    avg_loss = total_loss / total_samples
    avg_acc = correct_predictions / total_samples
    return avg_loss, avg_acc

# --- Define the function for one validation epoch ---
def validate_one_epoch(model, data_loader, criterion, device):
    model.eval() # Set the model to "evaluation mode"
    
    total_loss = 0.0
    correct_predictions = 0
    total_samples = 0
    
    # We don't calculate gradients for validation
    with torch.no_grad():
        for spectrograms, labels in tqdm(data_loader, desc="Validating", leave=False):
            # Move data to the GPU
            spectrograms = spectrograms.to(device)
            labels = labels.to(device)
            
            # 1. Forward pass
            outputs = model(spectrograms)
            
            # 2. Calculate the loss
            loss = criterion(outputs, labels)
            
            # --- Calculate running statistics ---
            total_loss += loss.item() * spectrograms.size(0)
            _, predictions = torch.max(outputs, 1)
            correct_predictions += (predictions == labels).sum().item()
            total_samples += labels.size(0)
            
    avg_loss = total_loss / total_samples
    avg_acc = correct_predictions / total_samples
    return avg_loss, avg_acc

print("Training and validation functions defined successfully.")
```

```python
# Cell 9: Run the Main Training Loop (with Correct Plotting)

# --- We need lists to store the results for plotting ---
history = {
    'train_loss': [],
    'train_acc': [],
    'val_loss': [],
    'val_acc': []
}

# --- Configuration ---
NUM_EPOCHS = 10  # We'll train for 10 full passes over the data
BEST_VAL_ACC = 0.0  # We'll use this to track our best model
MODEL_SAVE_PATH = "best_gru_encoder_model.pth" # Save file for the GRU model

print(f"--- Starting training for {NUM_EPOCHS} epochs on device: {device} ---")

for epoch in range(NUM_EPOCHS):
    start_time = time.time()
    
    # --- Train for one epoch ---
    train_loss, train_acc = train_one_epoch(
        model, 
        train_loader, 
        criterion, 
        optimizer, 
        device
    )
    
    # --- Validate for one epoch ---
    val_loss, val_acc = validate_one_epoch(
        model, 
        val_loader, 
        criterion, 
        device
    )
    
    end_time = time.time()
    
    # --- Store results for plotting ---
    history['train_loss'].append(train_loss)
    history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss)
    history['val_acc'].append(val_acc)
    
    # --- Print Epoch Results ---
    # (We will NOT clear the output)
    print(f"\nEpoch {epoch+1}/{NUM_EPOCHS} --- Time: {end_time-start_time:.2f}s")
    print(f"  Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}%")
    print(f"  Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc*100:.2f}%")
    
    # --- Save the best model ---
    if val_acc > BEST_VAL_ACC:
        print(f"  🎉 New best model! Validation accuracy improved from {BEST_VAL_ACC*100:.2f}% to {val_acc*100:.2f}%")
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
        BEST_VAL_ACC = val_acc
        print(f"  Model saved to {MODEL_SAVE_PATH}")
        
    print("-" * 50)

print(f"--- Training complete! ---")
print(f"Best validation accuracy achieved: {BEST_VAL_ACC*100:.2f}%")

# ---
# NEW PLOTTING CODE: This will run ONCE at the very end
# ---
print("\n--- Plotting Training History ---")
# Create an array for the x-axis (Epochs 1 to N)
epochs_range = range(1, NUM_EPOCHS + 1)

# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Loss
ax1.plot(epochs_range, history['train_loss'], 'o-', label='Train Loss') # 'o-' adds dots
ax1.plot(epochs_range, history['val_loss'], 'o-', label='Val Loss')
ax1.set_title(f'Training & Validation Loss')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()
ax1.grid(True)

# Plot 2: Accuracy
ax2.plot(epochs_range, history['train_acc'], 'o-', label='Train Acc') # 'o-' adds dots
ax2.plot(epochs_range, history['val_acc'], 'o-', label='Val Acc')
ax2.set_title(f'Training & Validation Accuracy')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.legend()
ax2.grid(True)

# Display the plot
plt.show()
```

```python
# Cell 11: Create the Embedding Extractor Model

class EmbeddingExtractor(SpeakerEncoder):
    """
    A modified version of SpeakerEncoder that stops after the embedding layer.
    It inherits the full CNN-GRU architecture but only returns the 768-dim vector.
    """
    def __init__(self, num_speakers):
        super(EmbeddingExtractor, self).__init__(num_speakers)

    def forward(self, x):
        # 1. Pass through CNN blocks
        x = F.relu(self.conv1(x))
        x = self.pool1(x)
        x = F.relu(self.conv2(x))
        x = self.pool2(x)
        x = F.relu(self.conv3(x))
        x = self.pool3(x)
        
        # 2. Reshape for GRU
        x = x.permute(0, 3, 2, 1)
        x = x.reshape(x.size(0), x.size(1), -1) 
        
        # 3. Pass through GRU and take final hidden state
        gru_output, hidden = self.gru(x)
        x = hidden.squeeze(0) 

        # 4. Calculate the 768-dim embedding (using the trained fc1_embedding layer)
        x_embedding = F.relu(self.fc1_embedding(x))
        
        # --- STOP HERE: Return the embedding vector ---
        return x_embedding

# --- Create and Load Weights ---

# 1. Create an instance of the new EmbeddingExtractor model
embedding_model = EmbeddingExtractor(num_speakers=num_speakers)

# 2. Copy the trained weights from loaded_model (Cell 10) into embedding_model
#    strict=False tells PyTorch to "not worry" about the missing
#    fc2_classifier layer, which is exactly what we want.
embedding_model.load_state_dict(loaded_model.state_dict(), strict=False) 

# 3. Move to GPU and set to evaluation mode
embedding_model.to(device)
embedding_model.eval()

print("EmbeddingExtractor (CNN-GRU) model created and weights loaded successfully.")

# --- Quick Test ---
print("\nTesting embedding model with one batch on GPU...")
try:
    with torch.no_grad():
        # (one_batch_spectrograms variable comes from Cell 5)
        test_batch_gpu = one_batch_spectrograms.to(device)
        embeddings = embedding_model(test_batch_gpu)
    
    # Expected output shape: [Batch Size, 768]
    print(f"✅ Success! Embedding model ran.")
    print(f"   Output embedding shape: {embeddings.shape} (Batch Size, 768)")

except Exception as e:
    print(f"❌ An error occurred during the embedding model test: {e}")
```

```python
# Cell 12: Extract Embeddings (LibriSpeech Diagnostic Test)

print("Starting VAD-less embedding extraction for diagnostic test...")
# --- 1. Define Audio File Path ---
meeting_audio_path = r"C:\Users\pavan\MEETINGDIARENC\data\LibriSpeech\train-clean-100\19\227\19-227-0000.flac"

# --- 2. Load the full audio ---
try:
    y_meeting, sr_meeting = librosa.load(meeting_audio_path, sr=16000, mono=True)
    print(f"Audio loaded: {meeting_audio_path}")
    print(f"Audio loaded: duration={len(y_meeting)/sr_meeting:.2f} seconds")
except FileNotFoundError:
    print(f"❌ ERROR: File not found: {meeting_audio_path}")
    raise

# --- 3. Define Sliding Window Parameters ---
window_len_s = 3.0
hop_len_s = 0.3     
window_len_samples = int(window_len_s * sr_meeting) 
hop_len_samples = int(hop_len_s * sr_meeting)     

# --- 4. Iterate with Sliding Window ---
embeddings_list = []
timestamps_list = []
audio_chunks_list = [] 

num_steps = int(np.ceil((len(y_meeting) - window_len_samples) / hop_len_samples)) + 1

with torch.no_grad():
    for i in tqdm(range(num_steps), desc="Processing Chunks"):
        start_sample = i * hop_len_samples
        end_sample = start_sample + window_len_samples
        y_chunk = y_meeting[start_sample:end_sample]
        
        if len(y_chunk) < window_len_samples:
            y_chunk = np.pad(y_chunk, (0, window_len_samples - len(y_chunk)), 'constant')

        S_mel = librosa.feature.melspectrogram(y=y_chunk, sr=sr_meeting, n_mels=80)
        S_mel_db = librosa.power_to_db(S_mel, ref=np.max)
        
        if S_mel_db.shape[1] > 94:
            S_mel_db = S_mel_db[:, :94]
        elif S_mel_db.shape[1] < 94:
            S_mel_db = np.pad(S_mel_db, ((0, 0), (0, 94 - S_mel_db.shape[1])), 'constant')
            
        spectrogram_tensor = torch.tensor(S_mel_db, dtype=torch.float32).unsqueeze(0).unsqueeze(0).to(device)
        
        embedding = embedding_model(spectrogram_tensor)
        embeddings_list.append(embedding.cpu().numpy().squeeze())
        
        timestamps_list.append({'start': start_sample / sr_meeting, 'end': (start_sample + window_len_samples) / sr_meeting})
        audio_chunks_list.append(np.mean(librosa.feature.rms(y=y_chunk)))

# --- 5. Stack Embeddings ---
all_embeddings = np.array(embeddings_list)
all_rms_values = np.array(audio_chunks_list)

print(f"\n✅ Successfully extracted {len(all_embeddings)} embeddings.")
```

```python
# Cell 13: Cluster Embeddings (1 Speaker)

def collapse_annotations(timestamps_list, labels, silence_labels_set, all_rms_values, rms_threshold=0.005, min_duration=0.5):
    hypothesis = Annotation()
    current_speaker = None
    current_start_time = 0
    labels_extended = list(labels) + [None] 
    rms_extended = list(all_rms_values) + [np.inf] 
    
    for i, label in enumerate(labels_extended):
        if i < len(timestamps_list):
            chunk_start_time = timestamps_list[i]['start']
        else: 
            chunk_start_time = timestamps_list[-1]['end']
        
        is_silence = False
        if i < len(labels):
            is_cluster_silence = label in silence_labels_set
            is_too_quiet = rms_extended[i] < rms_threshold
            if is_cluster_silence or is_too_quiet:
                is_silence = True
        
        new_speaker = label if not is_silence else -1
        if current_speaker is None:
            current_speaker = new_speaker
            current_start_time = chunk_start_time
            continue
        if new_speaker != current_speaker:
            segment_end_time = chunk_start_time
            if current_speaker != -1:
                segment = Segment(current_start_time, segment_end_time)
                if segment.duration > min_duration: 
                    hypothesis[segment] = f"Speaker_{current_speaker}"
            current_speaker = new_speaker
            current_start_time = chunk_start_time
    return hypothesis

# 1. Define Parameters
NUM_SPEAKERS = 1 
NUM_CLUSTERS = NUM_SPEAKERS + 1
print(f"Running Spectral Clustering, asking for {NUM_CLUSTERS} clusters...")

# 2. Cluster
clustering = SpectralClustering(n_clusters=NUM_CLUSTERS, affinity='cosine', assign_labels='kmeans', random_state=42).fit(all_embeddings)
labels = clustering.labels_ 

# 3. Find Silence
cluster_energies = [np.mean(all_rms_values[labels == i]) if len(all_rms_values[labels == i]) > 0 else np.inf for i in range(NUM_CLUSTERS)]
silence_labels_set = {np.argmin(cluster_energies)}
print(f"Identified 'silence' cluster: {silence_labels_set}")

# 4. Smooth
labels = medfilt(labels, kernel_size=5).astype(int)

# 5. Collapse
hypothesis = collapse_annotations(
    timestamps_list, labels, silence_labels_set, all_rms_values, 
    rms_threshold=0.005, min_duration=0.5
)
print(f"Total *non-overlapping* speech segments found: {len(list(hypothesis.itersegments()))}")
```

```python
# Cell 14: Final Evaluation with RTTM File

from pyannote.metrics.diarization import DiarizationErrorRate
from pyannote.core import Annotation, Segment
import os
import warnings
import pandas as pd
import io 

print("Imported libraries for RTTM parsing and DER...")

# --- IMPORTANT: PATH TO YOUR CUSTOM RTTM FILE ---
RTTM_FILE_PATH = r"C:\Users\pavan\MEETINGDIARENC\LIBRISPEECH_RAW_TEST.rttm"
# -------------------------------------------------

def load_rttm_reference(rttm_path):
    """Loads a reference PyAnnote Annotation object from the RTTM file."""
    reference = Annotation()
    try:
        with open(rttm_path, 'r') as f:
            for line in f:
                if line.startswith('SPEAKER'):
                    fields = line.strip().split()
                    start_time = float(fields[3])
                    duration = float(fields[4])
                    end_time = start_time + duration
                    speaker_label = fields[7]
                    time_segment = Segment(start_time, end_time)
                    reference[time_segment] = speaker_label
        print(f"Successfully loaded {len(list(reference.itersegments()))} reference segments from: {RTTM_FILE_PATH}")
    except FileNotFoundError:
        print(f"❌ ERROR: RTTM file not found at the specified path: {rttm_path}")
        print("Please make sure you have created this file with the correct RTTM content.")
        raise
    return reference

try:
    # Suppress the UserWarning from pyannote about 'uem'
    warnings.filterwarnings("ignore", category=UserWarning)

    # --- 1. Load the Ground Truth (Answer Key) ---
    reference = load_rttm_reference(RTTM_FILE_PATH)

    # --- 2. Calculate the DER ---
    # We use a 0.0s collar because this is a clean, simple file test
    der_metric = DiarizationErrorRate(collar=0.0) 
    
    # Compare the reference (RTTM) against the hypothesis (from Cell 13)
    final_der = der_metric(reference, hypothesis) 
    
    print("\n--- Diagnostic DER Report (Clean LibriSpeech) ---")
    # This prints the full report (Confusion, Missed Detection, etc.)
    print(der_metric.report(display=False))
    
    print("\n-------------------------------------------------")
    print(f"  Final DER = {final_der * 100:.2f}%")
    print("-------------------------------------------------")
    print("\n(A 'confusion' error of 0% means this test was a success.)")

except FileNotFoundError as e:
    print(f"❌ ERROR: {e}")
except NameError:
    print(f"❌ An error occurred. Did you run Cell 13 first to generate 'hypothesis'? The hypothesis variable is missing.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
